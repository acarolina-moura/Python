```python
diretores_count = df_movies['Director'].value_counts()
print(diretores_count)
top_3_diretores = diretores_count.head(3).index
```

```python
df_top_diretores = df_movies[df_movies['Director'].isin(top_3_diretores)]

resultado = df_top_diretores.groupby('Director').agg(
    Total_Filmes=('Title', 'count'),
    Soma_Box_Office=('Box Office', 'sum'),
    Soma_Budget=('Budget', 'sum'),
    Media_Run_Time=('Run Time', 'mean')
)
print(resultado)
```

```python
df_movies['Release Date'] = pd.to_datetime(df_movies['Release Date'], errors='coerce')

df_movies['Decade'] = (df_movies['Release Date'].dt.year // 10 * 10).astype(str) + 's'

resultado_decada_genero = df_movies.groupby(['Decade', 'Genre']).agg(
    Total_Filmes=('Title', 'count'),
    Media_Box_Office=('Box Office', 'mean')
)

print(resultado_decada_genero)
```
