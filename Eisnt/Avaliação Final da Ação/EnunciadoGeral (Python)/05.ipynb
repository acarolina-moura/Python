```python
# 5. Usando as bibliotecas Pandas e MatPlotLib e um dataset, elabora um notebook 
# jupyter no qual efetues: 

# • Limpeza e tratamento de dados; 
# • Processamento de dados: groupby, filter, criação de novas colunas,…; 
# • Visualização de dados;

import pandas as pd

df = pd.read_csv('dataset.csv')

print(df.head())

df.fillna(0, inplace=True)

df['Nova_Coluna'] = df['Idade'] + df['Salario']

agrupado = df.groupby('Departamento').agg({
    'Idade': 'mean', 
    'Salario': 'mean', 
    'Nova_Coluna': 'mean'
})

print(agrupado)
```
