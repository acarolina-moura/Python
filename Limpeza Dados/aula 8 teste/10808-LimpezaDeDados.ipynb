## Exemplo de NumPy

```python
import numpy as np

# Simulação de dados de vendas com valores ausentes (np.nan)
vendas = np.array([100, 200, np.nan, 400, 500, np.nan, 700])

# Identificação de valores nulos
valores_nulos = np.isnan(vendas)
print("Valores nulos encontrados:", valores_nulos)

# Substituição de valores nulos pela média dos valores existentes
media = np.nanmean(vendas)
vendas_limpo = np.where(np.isnan(vendas), media, vendas)
print("Dados após tratamento:", vendas_limpo)
```

## Exemplo de Machine Learning com Scikit-Learn

```python
from sklearn.linear_model import LinearRegression
import pandas as pd

# Dados de vendas
vendas = pd.DataFrame({
    'Publicidade': [1000, 1500, 2000, 2500],
    'Vendas': [200, 300, 400, 500]
})

# Modelo de regressão linear
modelo = LinearRegression()
modelo.fit(vendas[['Publicidade']], vendas['Vendas'])

# Previsão de vendas com base em publicidade
previsao = modelo.predict([[3000]])
print(f"Previsão de vendas: {previsao[0]:.2f} unidades")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

# Dataset com valores ausentes
dados = pd.DataFrame({
    'Idade': [25, 30, np.nan, 45, 50],
    'Salário': [50000, 60000, 55000, np.nan, 65000]
})

# Imputação de valores omissos com a média
imputador = SimpleImputer(strategy='mean')
dados[['Idade', 'Salário']] = imputador.fit_transform(dados[['Idade', 'Salário']])

# Normalização dos dados
escalador = StandardScaler()
dados[['Idade', 'Salário']] = escalador.fit_transform(dados[['Idade', 'Salário']])

print("Dados após pré-processamento:\n", dados)
```

## Filtragem de dados

```python
import pandas as pd

# Exemplo de DataFrame
df = pd.DataFrame({
    'Produto': ['Arroz', 'Feijão', 'Macarrão', 'Batata'],
    'Preço': [1.2, 2.5, 1.8, 0.9],
    'Stock': [100, 50, 70, 30]
})

# Filtrar produtos com preço maior que 1.5
df_filtrado = df[df['Preço'] > 1.5]
```

```python
df.mean()
```

```python
# Filtrar produtos com preço > 1.5 e stock > 50
df_filtrado = df[(df['Preço'] > 1.5) & (df['Stock'] > 50)]

# Filtrar produtos com preço < 1.0 ou stock < 40
df_filtrado = df[(df['Preço'] < 1.0) | (df['Stock'] < 40)]
```

```python
# Filtrar produtos específicos
produtos_selecionados = ['Arroz', 'Feijão']
df_filtrado = df[df['Produto'].isin(produtos_selecionados)]

# Filtrar produtos com preço entre 1.0 e 2.0
df_filtrado = df[df['Preço'].between(1.0, 2.0)]

# Filtrar produtos cujo nome contém 'a'
df_filtrado = df[df['Produto'].str.contains('a')]
```

```python

# Criar nova coluna com classificação de preço usando apply()
df['Categoria'] = df['Preço'].apply(lambda x: 'Alto' if x > 1.5 else 'Baixo')

# Substituir valores específicos com replace()
df['Produto'] = df['Produto'].replace('Arroz', 'Arroz Integral')

# Converter tipo de dados com astype()
df['Preço'] = df['Preço'].astype(float)

# Aplicar limites com clip() - Ajusta os valores sem remover linhas
df['Preço'] = df['Preço'].clip(lower=1.0, upper=2.0)

# Substituir valores com base em condição com where()
# Mantém os valores acima de 1.5 e substitui os restantes por 0
df['Desconto'] = df['Preço'].where(df['Preço'] > 1.5, other=0)

# Aplicar função com map() em uma Series
df['Produto'] = df['Produto'].map({'Arroz': 'Arroz Integral', 'Feijão': 'Feijão Preto'})
```

```python
# Filtrar colunas específicas
colunas_filtradas = df.filter(items=['Produto', 'Preço'])

# Filtrar colunas que contêm a palavra 'Pre'
colunas_like = df.filter(like='Pre', axis=1)

# Filtrar colunas com expressão regular (começam com 'P')
colunas_regex = df.filter(regex='^P', axis=1)
```

```python
# Query com múltiplas condições
categorias = ['Alimentos', 'Bebidas']
df.query('Categoria in @categorias and (Preço > 1.0 or stock < 50)')

# Expressões matemáticas e lógicas
fator = 1.2
df.query('(Preço * @fator) > 3.0 and stock >= 10')
```

```python
# Exemplo básico com condição simples
df.query('Preço > 1.5')

# Query com múltiplas condições
df.query('Preço > 1.5 and Categoria == "Alimentos"')

# Query com variáveis externas
limite = 2.0
df.query('Preço < @limite')
```

```python
# Exemplo de DataFrame
import pandas as pd

df = pd.DataFrame({
    'Price': [10, 20, 30, 40, None],
    'Stock': [5, 10, 15, 20, 25]
})

# Calcular a média dos preços
media_preco = df['Price'].mean()
print('Média do Preço:', media_preco)

# Calcular a soma do stock total
soma_stock = df['Stock'].sum()
print('Soma do Stock:', soma_stock)

# Contar quantos valores não nulos existem na coluna 'Price'
contagem_preco = df['Price'].count()
print('Contagem de Preço:', contagem_preco)

# Valor mínimo da coluna 'Price'
preco_minimo = df['Price'].min()
print('Preço Mínimo:', preco_minimo)

# Valor máximo da coluna 'Price'
preco_maximo = df['Price'].max()
print('Preço Máximo:', preco_maximo)

# Desvio padrão da coluna 'Price'
desvio_preco = df['Price'].std()
print('Desvio Padrão do Preço:', desvio_preco)

# Mediana da coluna 'Price'
mediana_preco = df['Price'].median()
print('Mediana do Preço:', mediana_preco)

# Resumo estatístico completo do DataFrame
resumo_estatistico = df.describe()
print('Resumo Estatístico Completo:\
', resumo_estatistico)

uniquos = df['Price'].nunique()
print(uniquos)
```

```python
df = pd.DataFrame({
    'Product': ['Rice', 'Beans', 'Pasta', 'Rice', 'Beans'],
    'Category': ['Food', 'Food', 'Food', 'Beverage', 'Food'],
    'Price': [1.5, 2.0, 1.8, 1.5, 2.0],
    'Stock': [100, 150, 80, 100, 150],
    'Store': ['Lisbon', 'Porto', 'Lisbon', 'Porto', 'Lisbon']
})

# Ordenar pelo preço de forma decrescente
df_sorted = df.sort_values(by='Price', ascending=False)
print('Ordenado por Preço (decrescente):', df_sorted)

# Agrupar por categoria e calcular a média de preço
df_grouped = df.groupby('Category')['Price'].mean()
print('Média de Preço por Categoria:', df_grouped)

# Criar tabela dinâmica da média de preço por loja e categoria
df_pivot = df.pivot_table(values='Price', index='Category', columns='Store', aggfunc='mean')
print('Tabela Dinâmica:', df_pivot)

# Definir a coluna 'Product' como índice
df_indexed = df.set_index('Product')
print(' DataFrame com índice definido:', df_indexed)

# Resetar o índice
df_reset = df_indexed.reset_index()
print('DataFrame com índice resetado:', df_reset)
```

## Combinação de Dataframes

```python
import pandas as pd

# Exemplo de DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nome': ['João', 'Maria', 'Carlos']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Idade': [25, 30, 35]
})

# Combinação com merge (INNER JOIN por padrão)
df_merged = pd.merge(df1, df2, on='ID')
print(df_merged)

# LEFT JOIN
df_left = pd.merge(df1, df2, on='ID', how='left')

# RIGHT JOIN
df_right = pd.merge(df1, df2, on='ID', how='right')

# OUTER JOIN
df_outer = pd.merge(df1, df2, on='ID', how='outer')
```

```python
# Exemplo de DataFrames
df3 = pd.DataFrame({
    'ID': [4, 5],
    'Nome': ['Ana', 'Pedro']
})

# Concatenação ao longo das linhas (axis=0)
df_concat = pd.concat([df1, df3], axis=0)
print(df_concat)
```

```python
import pandas as pd

# 1️⃣ Importação e Visualização de Dados

# Q1. Importar o ficheiro CSV
df = pd.read_csv('large_sales_data.csv')

# Q2. Função para mostrar as 5 primeiras linhas
def firstFive(dataframe):
    print("As primeiras 5 linhas do DataFrame disponibilizado são:")
    print(dataframe.head())

firstFive(df)

# Q3. Verificar o formato e estatísticas descritivas
print(df.info())
print(df.describe())

# Q4. Visualizar colunas Product e Price
print(df[['Product', 'Price']])

# 2️⃣ Manipulação de Colunas

# Q1. Criar coluna Desconto
df['Desconto'] = df['Price'] * 0.9

# Q2. Criar coluna Value_Stock
df['Value_Stock'] = df['Price'] * df['Stock']

# Q3. Renomear coluna Price para Unit_Price
df.rename(columns={'Price': 'Unit_Price'}, inplace=True)

# 3️⃣ Filtragem de Dados

# Q1. Filtrar produtos com preço superior a 1.5
print(df[df['Unit_Price'] > 1.5])

# Q2. Filtrar produtos com stock inferior a 50
print(df[df['Stock'] < 50])

# Q3. Filtrar produtos da categoria 'Food' com preço entre 1.0 e 2.0
print(df[(df['Category'] == 'Food') & (df['Unit_Price'].between(1.0, 2.0))])

# Q4. Filtrar produtos vendidos em Lisboa com stock superior a 60
print(df[(df['Store'] == 'Lisbon') & (df['Stock'] > 60)])

# Q5. Filtrar produtos cujo nome contém a palavra 'Rice'
print(df[df['Product'].str.contains('Rice')])

# 4️⃣ Utilização de Métodos

# Q1. Filtrar com query()
print(df.query("Category == 'Beverage' and Unit_Price < 2.0"))

# Q2. Filtrar produtos com a letra 'a' no nome
print(df[df['Product'].str.contains('a')])

# Q3. Filtrar categorias Dairy e Meat usando isin()
print(df[df['Category'].isin(['Dairy', 'Meat'])])

# Q4. Selecionar preços entre 1.0 e 2.0
print(df[df['Unit_Price'].between(1.0, 2.0)])

# Q5. Aumentar o Stock em 20 unidades para a categoria Food
df['Stock'] = df.apply(lambda x: x['Stock'] + 20 if x['Category'] == 'Food' else x['Stock'], axis=1)

# 5️⃣ Criação, Filtros e Métodos

# Q1. Criar coluna Classificação
df['Classificação'] = df['Unit_Price'].apply(lambda x: 'Caro' if x > 2.0 else 'Barato')
print(df[df['Classificação'] == 'Barato'])

# Q2. Criar coluna Reposição_Necessária
df['Reposição_Necessária'] = df['Stock'].apply(lambda x: 'Sim' if x < 40 else 'Não')
print(df[(df['Reposição_Necessária'] == 'Sim') & (df['Unit_Price'].between(1.0, 2.0))])

# 6️⃣ Ordenação e Análise de Dados

# Q1. Ordenar pelo Stock
df_sorted_stock = df.sort_values(by='Stock', ascending=False)
print(df_sorted_stock)

# Selecionar os 3 produtos mais caros
print(df.sort_values(by='Unit_Price', ascending=False).head(3))

# Q2. Calcular o preço médio por categoria
print(df.groupby('Category')['Unit_Price'].mean())

# Contar produtos por loja
print(df['Store'].value_counts())

# Q3. Pivot table para média de preço por categoria e loja
pivot = df.pivot_table(values='Unit_Price', index='Category', columns='Store', aggfunc='mean')
print(pivot)
```

## Outliers

```python
from scipy import stats
import numpy as np
import pandas as pd

# Criar dataset fictício
df = pd.DataFrame({'Vendas': [100, 101, 100, 3000, 102, 104, 150]})

# Calcular Z-Score
df['Z-Score'] = np.abs(stats.zscore(df['Vendas']))

# Filtrar outliers, Neste caso, 3 não seria adequado para remover os 3000
df_sem_outliers = df[df['Z-Score'] <= 3] 
display(df_sem_outliers)
```

```python
df
```

```python
Q1 = df['Vendas'].quantile(0.25)
Q3 = df['Vendas'].quantile(0.75)
IQR = Q3 - Q1

# Definir limites para outliers
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Remover outliers
df_sem_outliers = df[(df['Vendas'] >= limite_inferior) & (df['Vendas'] <= limite_superior)]
df_sem_outliers
```

```python
from sklearn.ensemble import IsolationForest

df = pd.DataFrame({'Vendas': [100, 101, 100, 3000, 102, 104, 150]})

iso_forest = IsolationForest(n_estimators=100, contamination='auto', random_state=42)
df['Outlier_IF'] = iso_forest.fit_predict(df[['Vendas']])

# Filtrar valores normais
df_sem_outliers = df[df['Outlier_IF'] == 1]
df_sem_outliers
```

```python
df
```

```python
from sklearn.neighbors import LocalOutlierFactor

df = pd.DataFrame({'Vendas': [100, 120, 110, 3000, 130, 140, 150]})

lof = LocalOutlierFactor(n_neighbors=5, metric='euclidean')
df['Outlier_LOF'] = lof.fit_predict(df[['Vendas']])

# Filtrar valores normais
df_sem_outliers = df[df['Outlier_LOF'] == 1]
df_sem_outliers
```

```python
print("Valores ausentes antes do tratamento:")
display(df.isnull().sum())

# Limpar espaços ou strings estranhas por NaN
df = df.replace(' ', np.nan)
df= df.replace("NaN", np.nan)
df = df.replace("", np.nan)


print("Valores ausentes após replace do tratamento:")
display(df.isnull().sum())
# Exemplo: preencher Horas_Extras com 0
if "Horas_Extras" in df.columns:
    df["Horas_Extras"] = df["Horas_Extras"].fillna(0)

# Exemplo: preencher Satisfação com a moda
if "Satisfação" in df.columns and df["Satisfação"].isnull().sum() > 0:
    moda_satisf = df["Satisfação"].mode()[0]
    df["Satisfação"].fillna(moda_satisf, inplace=True)

# Usar SimpleImputer para colunas numéricas (Idade, Salário)
from sklearn.impute import SimpleImputer

imputer_median = SimpleImputer(strategy="median")
df[["Idade", "Salário"]] = imputer_median.fit_transform(df[["Idade", "Salário"]])

# Colunas categóricas => "Desconhecido"
df[["Departamento","Região"]] = df[["Departamento","Região"]].fillna("Desconhecido")

print("\nValores ausentes após tratamento:")
display(df.isnull().sum())
```

### Remoção de outliers

```python
print("Linhas antes:", df.shape)

# Definição de limites lógicos
if "Idade" in df.columns:
    # Eliminar linhas onde Idade > 100
    df = df[(df["Idade"] <= 100) & (df["Idade"] > 0)]

if "Salário" in df.columns:
    # Eliminar linhas onde Salário < 500 ou Salário > 30000
    df = df[(df["Salário"] >= 0) & (df["Salário"] <= 30000)]

print("Linhas com outliers removidas. Dimensões agora:", df.shape)
```

```python
# Z-Score
from scipy import stats
# Selecionar as colunas numéricas
numeric_cols = df.select_dtypes(include=[np.number]).columns

print("Colunas numéricas para analisar outliers (Z-Score):")
print(numeric_cols)


zscore_threshold = 3.0  # padrão
df_z = df[numeric_cols].apply(stats.zscore)  # DataFrame com o z-score de cada coluna

# Determinar quais linhas são outliers em pelo menos uma coluna
outliers_mask_z = (df_z.abs() > zscore_threshold).any(axis=1)

# Separar os outliers e os não outliers
df_outliers_z = df[outliers_mask_z]
df = df[~outliers_mask_z]

print(f"\nNº de outliers (Z-Score) encontrados: {df_outliers_z.shape[0]}")
print(f"Nº de linhas não outliers (resto do dataset): {df.shape[0]}")
```

```python
import numpy as np

print("Linhas antes do quartil:", df.shape)


q1 = df['Salário'].quantile(0.25)
q3 = df['Salário'].quantile(0.75)
iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr


print("Linhas com outliers removidas. Dimensões agora:", df.shape)

df = df[(df['Salário'] >= limite_inferior) & (df['Salário'] <= limite_superior)]
```

```python
from sklearn.ensemble import IsolationForest

# Selecionar colunas numéricas
numeric_cols = df.select_dtypes(include=[np.number]).columns

print("Colunas numéricas para analisar outliers (IsolationForest):")
print(numeric_cols)

iforest = IsolationForest(
    n_estimators=100,
    contamination=0.01,  # Exemplo: assume ~1% de contaminação
    random_state=42
)

iforest.fit(df[numeric_cols])

# Prever outliers: -1 = outlier, 1 = normal
y_pred = iforest.predict(df[numeric_cols])
outliers_mask_iforest = (y_pred == -1)

df_outliers_if = df[outliers_mask_iforest]

print(f"\nNº de outliers (IsolationForest) encontrados: {df_outliers_if.shape[0]}")
display(df_outliers_if)

```

### Codificação de Variáveis Categóricas

```python
from sklearn.preprocessing import OneHotEncoder


if "Departamento" in df.columns and "Região" in df.columns:
    encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
    colunas = ["Departamento", "Região"]

    encoded = encoder.fit_transform(df[colunas])
    nomes_novos = encoder.get_feature_names_out(colunas)

    df_encoded = pd.DataFrame(encoded, columns=nomes_novos, index=df.index)
    df = pd.concat([df, df_encoded], axis=1)

    print("One-Hot aplicado a 'Departamento' e 'Região'. Novas colunas:")
    print(list(nomes_novos))
else:
    print("Colunas categóricas não encontradas ou já tratadas.")


df.drop(columns=["Departamento", "Região"], inplace=True)
display(df)
```
