```python
import pandas as pd
import numpy as np

df = pd.read_excel("data.xlsx")
display (df.head(10))
```

```python
# 3. Verifique se existem registos duplicados e, caso existem, remova-os 

print("Verificação de duplicados:", df.duplicated().sum())
```

```python
# 4. Aplique técnicas apropriadas para a imputação de dados 

# Imputar valores numéricos com a média
df['Encomenda_Mais_Recente'] = pd.to_numeric(df['Encomenda_Mais_Recente'], errors='coerce')
df['Encomenda_Mais_Recente'] = df['Encomenda_Mais_Recente'].fillna(df['Encomenda_Mais_Recente'].mean())

df['Quantidade_Stock'] = pd.to_numeric(df['Quantidade_Stock'], errors='coerce')
df['Quantidade_Stock'] = df['Quantidade_Stock'].fillna(df['Quantidade_Stock'].mean())

df['Preço_Unitário'] = pd.to_numeric(df['Preço_Unitário'], errors='coerce')
df['Preço_Unitário'] = df['Preço_Unitário'].fillna(df['Preço_Unitário'].mean())

df['Preço_Custo'] = pd.to_numeric(df['Preço_Custo'], errors='coerce')
df['Preço_Custo'] = df['Preço_Custo'].fillna(df['Preço_Custo'].mean())

df['Unidades_Vendidas_Mensal'] = pd.to_numeric(df['Unidades_Vendidas_Mensal'], errors='coerce')
df['Unidades_Vendidas_Mensal'] = df['Unidades_Vendidas_Mensal'].fillna(df['Unidades_Vendidas_Mensal'].mean())

df['Quantidade_Encomendada'] = pd.to_numeric(df['Quantidade_Encomendada'], errors='coerce')
df['Quantidade_Encomendada'] = df['Quantidade_Encomendada'].fillna(df['Quantidade_Encomendada'].mean())


# Imputar valores categóricos com a moda
df['Tipo_Produto'] = df['Tipo_Produto'].fillna(df['Tipo_Produto'].mode()[0])
df['Armazém'] = df['Armazém'].fillna(df['Armazém'].mode()[0])
df['Região_Venda'] = df['Região_Venda'].fillna(df['Região_Venda'].mode()[0])
df['Satisfação_Clientes'] = df['Satisfação_Clientes'].fillna(df['Satisfação_Clientes'].mode()[0])
df['Nome_Produto'] = df['Nome_Produto'].fillna("Desconhecido")

print(f"Após imputação: {df.shape}")
```

```python
# 5. Aplique técnicas apropriadas para a remoção de valores extremos (outliers). Tendo também em conta as seguintes regras:

# Converter a coluna "Encomenda_Mais_Recente" para datetime, tratando erros
df['Encomenda_Mais_Recente'] = pd.to_datetime(df['Encomenda_Mais_Recente'], errors='coerce', unit='D', origin='1899-12-30')

produtos_grandes = df[(df["Tipo_Produto"] == "Grande") & (df["Quantidade_Stock"] > 500)]
produtos_medios = df[(df["Tipo_Produto"] == "Médio") & (df["Quantidade_Stock"] > 1000)]
produtos_pequenos = df[(df["Tipo_Produto"] == "Pequeno") & (df["Quantidade_Stock"] > 10000)]

df = df.drop(produtos_grandes.index)
df = df.drop(produtos_medios.index)
df = df.drop(produtos_pequenos.index)

df = df[df["Encomenda_Mais_Recente"].dt.year >= 1995]

print(f"após remoção de outliers: {df.shape}")
```

```python
df['Custo_Total_Stock'] = df['Preço_Unitário'] * df['Quantidade_Stock']

ano_atual = pd.Timestamp.now().year
df['Tempo_em_Stock'] = ano_atual - df['Encomenda_Mais_Recente'].dt.year

df['Margem_Lucro(%)'] = ((df['Preço_Unitário'] - df['Preço_Custo']) / df['Preço_Unitário']) * 100

print(df.head()) 
```

```python
# Média dos preços de custo por produto
media_preco_custo_por_produto = df.groupby('Nome_Produto')['Preço_Custo'].mean()
print("\nMédia dos preços de custo por produto:\n", media_preco_custo_por_produto)

contagem_produtos_por_armazem = df['Armazém'].value_counts()
print("\nContagem total de produtos por armazém:\n", contagem_produtos_por_armazem)

correlacao_stock_preco = df['Quantidade_Stock'].corr(df['Preço_Unitário'])
print("\nCorrelação entre quantidade em stock e preço de venda:\n", correlacao_stock_preco)

tabela_dinamica = pd.pivot_table(df, values='Custo_Total_Stock', index='Tipo_Produto', columns='Região_Venda', aggfunc='sum')
print("\nTabela dinâmica com o total de venda por tipo de produto e região de venda:\n", tabela_dinamica)


lucro_total = df['Custo_Total_Stock'].sum() * (df['Margem_Lucro(%)'].mean() / 100) 
print(f"\nLucro total estimado: {lucro_total:.2f}")

estatisticas_descritivas = df.describe()
print("\nEstatísticas descritivas para colunas numéricas:\n", estatisticas_descritivas)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler

df = pd.read_excel('data.xlsx')

print("Colunas:", df.columns.tolist())


scaler_std = StandardScaler()
df['Preco_Unitario_padronizado'] = scaler_std.fit_transform(df[['Preço_Unitário']])

scaler_minmax = MinMaxScaler()
df['Quantidade_Stock_normalizado'] = scaler_minmax.fit_transform(df[['Quantidade_Stock']])

print("Após Q8 (normalização/padronização):", df.shape)
print(df[['Preco_Unitario_padronizado', 'Quantidade_Stock_normalizado']].head())

```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
df['Tipo_Produto_encoded'] = label_encoder.fit_transform(df['Tipo_Produto'])

df = pd.get_dummies(df, columns=['Armazém', 'Região_Venda'])

print("Após Q9 (encoding):", df.shape)
print(df.head())
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.decomposition import PCA

colunas_pca = ['Quantidade_Stock', 'Quantidade_Encomendada', 'Preço_Unitário', 'Preço_Custo', 'Unidades_Vendidas_Mensal']
df[colunas_pca] = df[colunas_pca].apply(pd.to_numeric, errors='coerce')


imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(df[colunas_pca])


pca = PCA(n_components=2)
pca_result = pca.fit_transform(X_imputed)


df['PCA_1'] = pca_result[:, 0]
df['PCA_2'] = pca_result[:, 1]

print("Após aplicação do PCA:")
print(df[['PCA_1', 'PCA_2']].head())
```
