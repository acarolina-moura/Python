```python
from ipywidgets import interact
import math

def volume_esfera(raio):
    if raio < 0:
        return "O raio não pode ser negativo."
    volume = (4/3) * math.pi * (raio ** 3)
    return f"O volume da esfera é: {volume:.2f} unidades cúbicas"

interact(volume_esfera, raio=(0.0, 100.0, 0.1))
```
